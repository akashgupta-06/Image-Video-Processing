```python
import cv2
girl_img = cv2.imread(r'C:\Users\USER\Pictures\Saved Pictures\girl.jpg')

while True:                                    # using loop to close window automatically
    cv2.imshow('gril', girl_img)               # if read than show imag 

    if cv2.waitKey(1) & 0xFF == 27 :           # 
         break
cv2.destroyAllWindows()                  # No arguments required
```

```python
cv2.imwrite('girl.jpg',girl_img)
```
