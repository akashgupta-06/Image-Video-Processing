# real-time webcam video capture program using OpenCV 

## Black and Wight Video

```python
import cv2

cap = cv2.VideoCapture(0)

width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

while True:                                               # Continuously captures frames from the webcam.
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)         # Converts the color frame (BGR) to grayscale using OpenCV.

    cv2.imshow('frame',gray)                              # Displays the grayscale video in a window named "frame"

    if cv2.waitKey(1) & 0xFF == ord('q') :               #  press the 'q' key, it breaks the loop and stops the webcam.
        break
cap.release()
cv2.destroyAllWindows()
```

## Colored video

```python
import cv2

cap = cv2.VideoCapture(0)

width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

while True:
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)             #  Converts the color frame BGR to RGB using OpenCV.

    cv2.imshow('frame',gray)

    if cv2.waitKey(1) & 0xFF == ord('q') :
        break
cap.release()
cv2.destroyAllWindows()
```
