## Changing colour of image

```python
plt.imshow(horse_red[:,:,0])
plt.show()                                 # show in red
```

```python
plt.imshow(horse_red[:,:,1])             # green
plt.show()
```

```python
plt.imshow(horse_red[:,:,2])             # blue
plt.show()
```

```python
plt.imshow(horse_red[:,:,0],cmap = 'Greys')         # grey color
plt.show()
```

```python
plt.imshow(horse_red[:,:,0],cmap = 'grey')     # black and white
plt.show()
```

```python
plt.imshow(horse_red[:,:,1],cmap = 'YlOrBr')             # yellow  orange and brown
plt.show()
```

```python
plt.imshow(horse_red[:,:,1],cmap = 'Greens')            
plt.show()
```

## Removing red ,green and blue color from image

```python
horse_red[:,:,0]
```

```python
horse_red[:,:,1]
```

```python
horse_red[:,:,2]
```

```python
horse_red[:,:,0] = 0     # It sets the Red channel of every pixel in the image to 0
```

```python
horse_red[:,:,0]
```

```python
plt.imshow(horse_red)      # The image will lose all red color.
plt.show()                 # Only the green and blue channels remain active.
```

```python
horse_red[:,:,1] = 0
```

```python
horse_red[:,:,1]
```

```python
plt.imshow(horse_red)     # The image will lose all red and green color.
plt.show()                # image in blue colour
```

```python
horse_red[:,:,2] = 0           
```

```python
horse_red[:,:,2]                 
```

```python
plt.imshow(horse_red)                # all colour lost 
plt.show()                           # RGB (0, 0, 0) = Black
```
