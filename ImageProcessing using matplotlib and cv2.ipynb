## Resize the image

```python
fix_img_1 = cv2.resize(fix_img,(205,306))            # resize the img
```

```python
fix_img_1.shape
```

```python
w_ratio = 0.5
h_ratio = 0.5
```

```python
fix_img2 = cv2.resize(fix_img,(0,0), fix_img,w_ratio, h_ratio)
```

```python
fix_img2
```

```python
plt.imshow(fix_img2)
plt.show()
```

## Flip and Rotate image

```python
img3 = cv2.flip(fix_img2,0)    # Flip vertically (upside down)        
plt.imshow(img3)
plt.show()
```

```python
img3 = cv2.flip(fix_img2,1)     # Flip horizontally (left-right)
plt.imshow(img3)
plt.show()
```

```python
img3 = cv2.flip(fix_img2,-1)
plt.imshow(img3)                 # Flip both vertically and horizontally
plt.show()
```

```python
# rotate
img3 = cv2.rotate(fix_img2,0)
plt.imshow(img3)                 # 
plt.show()
```

```python
img3 = cv2.rotate(fix_img2,1)
plt.imshow(img3)                 # 
plt.show()
```

```python
img3 = cv2.rotate(fix_img2,2)
plt.imshow(img3)                 # 
plt.show()
```

## save image

```python
cv2.imwrite('elephant.jpg',img3)
```
